```python
import jax.numpy as jnp
from OkadaJAX import SPOINT, SRECTF
import matplotlib.pyplot as plt
plt.rcParams["font.size"] = 20

var_list = [
    "ux", "uy", "uz", "uxx", "uxy", "uyx", "uyy",  "uzx",  "uzy"
]
```

# `SPOINT` (single station)

```python
ALP = 0.5
X = jnp.array(0.5)
Y = jnp.array(-0.5)

D = 1.0
dip = jnp.array(45.0)
SD = jnp.sin(jnp.deg2rad(dip))
CD = jnp.cos(jnp.deg2rad(dip))

DISL1, DISL2, DISL3 = 4.0, 3.0, 0.0

out = SPOINT(ALP, X, Y, D, SD, CD, DISL1, DISL2, DISL3, compute_strain=True)
out
```

# `SPOINT` (multiple stations)

```python
ALP = 0.5
x = jnp.linspace(-1, 1, 101)
y = jnp.linspace(-1, 1, 101)
X, Y = jnp.meshgrid(x, y)


D = 1.0
dip = jnp.array(45.0)
SD = jnp.sin(jnp.deg2rad(dip))
CD = jnp.cos(jnp.deg2rad(dip))

DISL1, DISL2, DISL3 = 4.0, 3.0, 0.0

out = SPOINT(ALP, X, Y, D, SD, CD, DISL1, DISL2, DISL3, compute_strain=True)
out
```

```python
for i, var in enumerate(var_list):

    fig, ax = plt.subplots()
    im = ax.pcolormesh(X, Y, out[i], cmap="bwr", vmin=-0.5, vmax=0.5)
    ax.set_aspect("equal")
    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    ax.set_title(var, fontsize=28)
    fig.colorbar(im)
    fig.show()
```
